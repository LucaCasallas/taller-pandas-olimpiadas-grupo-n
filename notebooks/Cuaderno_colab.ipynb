## Fusión de ambos data frames
Utilizando .merge(), se han fusionado ambos data frames tomando como referencia la variable 'NOC', que se encuentra en ambos data frames.

```python
# FUSIÓN AMBOS DATASETS
df_fusion = pd.merge(atletas, regiones , on='NOC', how='left')
df_fusion.sort_values(by= 'ID' , ascending= True)
```

### ¿Cuántas filas quedan sin region? ¿Por qué cree que ocurre esto?
Por que en el dataframe de regiones, los valores 'TUV', 'ROT' y 'UNK' no tienen un valor en la columna 'NOC' por lo que llenan con NaN.

### ¿Qué tipo de fusión 'how' se usó? ¿Porque?
Usamos how='left'. Este tipo de fusión garantiza que el df de la izquierda(atletas) mantenga intactas todas sus filas incluso si algún atleta tiene un codigo NOC que no aparece en regiones.
##### Además, se descartan aquellos valores que se encuentren en el data frame 'regiones' que no tengan coincidencia con ninguna fila del data frame 'atletas'.

## Nueva columna: Medalla
Haciendo uso de .apply(), creamos la columna 'Medalla', que indica si un atleta ganó alguna medalla (ya sea de oro, plata o bronce). Esto lo hicimos limpiando los valores nulos de la columna 'Medal' y creando la función 'validarMedalla'.

```python
# LIMPIEZA DE LA COLUMNA "MEDAL"
df_fusion['Medal'] = df_fusion['Medal'].fillna('Ninguna')
df_fusion
```

```python
# CREACIÓN COLUMNA "MEDALLA"
def validarMedalla(x):
  if x == 'Ninguna':
    return 0
  return 1
```

```python
df_fusion['Medalla'] = df_fusion['Medal'].apply(validarMedalla)
df_fusion
```

## Gráficos

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
import numpy as np

d = df_fusion[df_fusion['IMC'].between(12, 50)]
plt.figure(figsize=(8, 4.5))
sns.histplot(data=d, x='IMC', hue='Sex', bins=60, stat='density', common_norm=False, element='step')
plt.title('Distribución del IMC por sexo'); plt.xlabel('IMC (kg/m²)')
plt.savefig('g01_imc_sexo.png', dpi=150, bbox_inches='tight'); plt.show()


deportes = ['Gymnastics', 'Athletics', 'Swimming', 'Football', 'Basketball',
            'Rowing', 'Judo', 'Boxing', 'Wrestling', 'Weightlifting']
d2 = d[d['Sport'].isin(deportes)]
orden = d2.groupby('Sport')['IMC'].median().sort_values().index
plt.figure(figsize=(9, 5))
sns.boxplot(data=d2, y='Sport', x='IMC', order=orden)
plt.title('IMC por deporte'); plt.ylabel('')
plt.savefig('g02_imc_deporte.png', dpi=150, bbox_inches='tight'); plt.show()


top = df_fusion.groupby('region')['Medalla'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(8, 4.5))
sns.barplot(x=top.values, y=top.index, color='#4c72b0')
plt.title('10 regiones con más registros con medalla'); plt.xlabel('Registros con medalla'); plt.ylabel('')
plt.savefig('g03_top_regiones.png', dpi=150, bbox_inches='tight'); plt.show()

t = df_fusion.groupby('Sex')['Medalla'].mean().reset_index()
plt.figure(figsize=(5, 4))
ax = sns.barplot(data=t, x='Sex', y='Medalla')
ax.bar_label(ax.containers[0], labels=[f'{v*100:.1f}%' for v in t['Medalla']])
plt.title('Tasa de medalla por sexo'); plt.ylabel('Medallas / participaciones')
plt.savefig('g04_tasa_sexo.png', dpi=150, bbox_inches='tight'); plt.show()

t = df_fusion.groupby('Season')['Medalla'].mean().reset_index()
plt.figure(figsize=(5, 4))
ax = sns.barplot(data=t, x='Season', y='Medalla')
ax.bar_label(ax.containers[0], labels=[f'{v*100:.1f}%' for v in t['Medalla']])
plt.title('Tasa de medalla por temporada'); plt.ylabel('Medallas / participaciones')
plt.savefig('g05_tasa_temporada.png', dpi=150, bbox_inches='tight'); plt.show()

d3 = d.copy()
d3['Resultado'] = d3['Medalla'].map({0: 'Sin medalla', 1: 'Con medalla'})
plt.figure(figsize=(7, 4.5))
sns.boxplot(data=d3, x='Sex', y='IMC', hue='Resultado')
plt.title('IMC según sexo y resultado')
plt.savefig('g06_imc_medalla.png', dpi=150, bbox_inches='tight'); plt.show()

tmp = df_fusion.assign(Periodo=np.where(df_fusion['Year'] < 1960, 'Antes de 1960', '1960 o después'))
falt = (tmp.groupby('Periodo')[['Age', 'Height', 'Weight', 'IMC']]
          .apply(lambda g: g.isna().mean())
          .reset_index()
          .melt(id_vars='Periodo', var_name='Variable', value_name='Proporcion'))

plt.figure(figsize=(7.5, 4.2))
ax = sns.barplot(data=falt, x='Variable', y='Proporcion', hue='Periodo', palette=['#c44e52', '#4c72b0'])
for c in ax.containers:
    ax.bar_label(c, labels=[f'{v*100:.1f}%' for v in c.datavalues], fontsize=8)
plt.title('Proporción de datos faltantes según el periodo')
plt.ylabel('Proporción de faltantes'); plt.xlabel('')
plt.savefig('g08_faltantes_periodo.png', dpi=150, bbox_inches='tight')
plt.show()


base = df_fusion.groupby(['Year', 'Season', 'Sex'])['ID'].nunique().reset_index(name='Atletas')

fig, ax = plt.subplots(1, 2, figsize=(13, 4.2))
for a, temporada in zip(ax, ['Summer', 'Winter']):
    sns.lineplot(data=base[base['Season'] == temporada], x='Year', y='Atletas',
                 hue='Sex', marker='o', ax=a, palette={'M': '#1f77b4', 'F': '#d62728'})
    a.set_title('Juegos de Verano' if temporada == 'Summer' else 'Juegos de Invierno')
    a.set_xlabel('Año')
    a.set_ylabel('Atletas únicos')
fig.suptitle('Participación por edición y sexo', y=1.02, fontsize=13)
plt.tight_layout()
plt.savefig('g07_participacion_anio_sexo.png', dpi=150, bbox_inches='tight')
plt.show()
```
